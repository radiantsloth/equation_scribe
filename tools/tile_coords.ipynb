```python
from PIL import Image, ImageDraw

img_path = "c:/data/repos/equation_scribe/detector/data/images/tiles_train/paper000_page_0000_tile_0000.png"
label_path = "c:/data/repos/equation_scribe/detector/data/labels/tiles_train/paper000_page_0000_tile_0000.txt"

im = Image.open(img_path)
draw = ImageDraw.Draw(im)
W, H = im.size

with open(label_path) as f:
    for line in f:
        if not line.strip(): continue
        cls, x, y, w, h = map(float, line.strip().split())
        
        x0 = (x - w/2) * W
        y0 = (y - h/2) * H
        x1 = (x + w/2) * W
        y1 = (y + h/2) * H
        
        # Draw rectangle (Red for class 0, Blue for class 1)
        color = "red" if int(cls) == 0 else "blue"
        draw.rectangle([x0, y0, x1, y1], outline=color, width=3)

im.show()
```
